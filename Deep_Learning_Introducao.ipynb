# Deep Learning Introdução

```python
import matplotlib as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score

# Coleta de dados (datasets) do sklearn
# https://scikit-learn.org/stable/api/sklearn.datasets.html
from sklearn import datasets

# Keras https://keras.io/
from keras.models import Sequential

# Lib para estruturar as camadas ocultas
from keras.layers import Dense

# Tensorflow https://www.tensorflow.org/?hl=pt-br
from tensorflow.keras.utils import to_categorical

import numpy as np
import pandas as pd
```

## Dataset

```python
# Carrega do dataset do sklearn
base = datasets.load_iris()
base
```

## Coletando previsores

```python
previsores = base.data
previsores
```

## Coletando classe

```python
classe = base.target
classe
```

## Representação Dummy

```python
# Classe 0 - [1, 0, 0]
# Classe 1 - [0, 1, 0]
# Classe 2 - [0, 0, 1]

classe_dummy = to_categorical(classe)
classe_dummy
```

## Modelo sequencial

```python
modelo = Sequential() # Escolha do modelo sequencial
```

### Criando modelo de camadas


#### Primeira camada (entrada)

```python
# primeira camada com 5 neuronios e 4 valores de entrada (previsores)
modelo.add(Dense(input_dim = 4, units = 5))
```

#### Segunda camada (camada oculta)

```python
# Camada oculta com 4 neuronios para afunilar (antes tinha 5)
modelo.add(Dense(units = 4))
```

#### Última camada (saida)

```python
# Camada com 3 neurônios, pois existem 3 possibilidades de plantas (Iris-setosa, Iris-versicolor ou Iris-virginica)
modelo.add(Dense(units = 3))
```

### Review do modelo

```python
modelo.summary() # Qtd de parametros que serao cruzados e quantos vao ser cruzados por camada
```

### Compilacao do modelo

```python
modelo.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

## Separação de dados para teste e treino

```python
X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(previsores,
                                                                  classe_dummy,
                                                                  test_size=0.3,
                                                                  random_state=0)
```

### Treinamento

```python
modelo.fit(X_treinamento,
           y_treinamento,
           epochs=500,
           validation_data=(X_teste, y_teste))
```

### Predição

```python
previsoes = modelo.predict(X_teste)
```

```python
# Converter os valores de probabilidade em 0 e 1
previsoes = (previsoes > 0.5)
previsoes
```

### Matriz de confusao

```python
# np.argmax -> pega a maior probabilidade para definir qual é a classe prevista
y_teste_matrix = [np.argmax(t) for t in y_teste]
y_previsoes_matrix = [np.argmax(t) for t in previsoes]
```

```python
confusao = confusion_matrix(y_teste_matrix, y_previsoes_matrix)
confusao
```

### Acuracia

```python
acuracia = accuracy_score(y_teste_matrix, y_previsoes_matrix)
acuracia
```
