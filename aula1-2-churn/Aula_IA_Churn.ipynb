# Tratamento de Dados

### Tratamento de dados ausentes/nulos


```python

dataset.isnull().sum() # Busca por dados nulos e soma
```

#### Genero

```python
# Dados textuais - Aplicar a MODA

dataset['Genero'].fillna('Masculino', inplace=True) # Buscar pelos valores nulos e preencher o o valor recebido por parametro
```

```python
dataset['Genero'].isnull().sum() # Validar que o tratamento deu certo (deve retornar 0)
```

#### Salario

```python
mediana = sts.median(dataset['Salario'])
mediana
```

```python
dataset['Salario'].fillna(mediana, inplace=True)
```

```python
dataset['Salario'].isnull().sum()
```

### Tratamento de dados errados

### Genero

```python
dataset.loc[dataset['Genero'] == 'M', 'Genero'] = 'Masculino' # Localiza valores dentro do dataset e substitui por o valor recebido
```

```python
dataset.loc[dataset['Genero'].isin(['Fem', 'F']), 'Genero'] = 'Feminino' # Localiza valores dentro do dataset e substitui por o valor recebido
```

```python
agrupado = dataset.groupby(['Genero']).size()
agrupado
```

#### Idade

```python
mediana = sts.median(dataset['Idade']) # Calcular a mediana de idade
mediana
```

```python
dataset.loc[(dataset['Idade'] < 0) |  (dataset['Idade'] > 116), 'Idade'] = mediana # Alterar todas as idades erradas, preenchendo com o valor da mediana
```

```python
dataset.loc[(dataset['Idade'] < 0) |  (dataset['Idade'] > 116), 'Idade']
```

#### Estados


```python
dataset.loc[dataset['Estado'].isin(['RP', 'SP', 'TD']), 'Estado'] = 'RS' # APlicar a MODA nos estados
agrupado = dataset.groupby(['Estado']).size()
agrupado
```

#### Registros duplicados

```python
dataset[dataset.duplicated(['ID'], keep=False)] # Busca registros duplicados
```

```python
dataset.drop_duplicates(subset='ID', keep='first', inplace=True) # Exclui valores duplicados
```

```python
dataset[dataset.duplicated(['ID'], keep=False)] # Validar que foi excluido os registros
```

#### Salario

```python
desvio = sts.stdev(dataset['Salario'])
desvio
```

```python
dataset.loc[dataset['Salario'] > 2 * desvio, 'Salario'] = mediana # Substitui os salarios que sao 2 vezes maior que o desvio
```

```python
dataset.loc[dataset['Salario'] > 2 * desvio, 'Salario'] # Valida se deu certo
```
