# Machine Learning

```python
from sklearn.model_selection import train_test_split # Dividir os dados entre treino e teste
from sklearn.metrics import accuracy_score, confusion_matrix # Avaliação do modelo com precisão e matriz de confusão
from sklearn.neighbors import KNeighborsClassifier # Algoritmo de KNN - Classificação por vizinhos (aproximação)
import pandas as pd
from scipy import stats
```

```python
dataset = pd.read_csv('iris.csv', sep=',')
```

```python
dataset.head()
```

```python
dataset.shape
```

```python
dados = pd.DataFrame(dataset) # Transformar dados em dataframe
```

```python
dados.describe() # Analise e estatisticas dos dados
```

```python
dados['class'].value_counts() # Ver quantos registros tem de cada planta, para saber se o treinamento faz sentido

# Precisa ter uma quantidade equalizada de dados
```

## Separação dos dados

```python
previsores = dados.iloc[:,0:4].values # Seleciona todas os valores, das colunas de 0 até 4 (todas as colunas, exceto a classe)
classe = dados.iloc[:,4].values # Seleciona todos os valores da coluna 4 (classe)
```

### Separação 70/30

```python
X_treinamento, X_teste, Y_treinamento, Y_teste = train_test_split(previsores, classe, test_size=0.3, random_state=0)

# Separação dos dados de 70% para treinamento, e 30% para testes
```

```python
len(X_treinamento) # Validar se a separacao esta correta (tem que retornar 105)
```

## Treinar o algoritmo

```python
knn = KNeighborsClassifier(n_neighbors=3) # Escolhe o algoritmo KNN e quantidade de vizinhos (3)
```

```python
knn.fit(X_treinamento, Y_treinamento) # Função para treinar o modelo com dados
```

```python
prova = knn.predict(X_teste) # Função para prever/testar o modelo
```

```python
prova # Respostas
```

## Calcular quantidade de acertos

```python
acertos = accuracy_score(Y_teste, prova)
acertos
```

### Matriz de confusão

```python
confusao = confusion_matrix(Y_teste, prova)
confusao
```
